```julia
using Plots

x = 0:0.1:10

plot(
    x,
    [sin.(x), cos.(x)],
    label=["sin(x)" "cos(x)"],
    title="Two functions",
    xlabel="x",
    ylabel="y"
)
```

```julia
using Plots

x = -5:0.2:5
y = -5:0.2:5

z = [x^2 + y^2 for y in y, x in x]

surface(x, y, z)
```

```julia
using Plots

x = -5:0.2:5
y = -5:0.2:5

z = [sin(sqrt(x^2 + y^2)) for y in y, x in x]

surface(x, y, z)
```
