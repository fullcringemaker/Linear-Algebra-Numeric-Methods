```julia
using PyPlot
```

```julia
x = range(0; stop=2*pi, length=1000)
y = sin.(3 * x + 4 * cos.(2 * x))

PyPlot.plot(
    x,
    y,
    color="red",
    linewidth=2.0,
    linestyle="--"
)

PyPlot.title("A sinusoidally modulated sinusoid")
```

```julia
using PyPlot

x = 0:0.05:2*pi

PyPlot.plot(x, sin.(x), label="sin(x)")
PyPlot.plot(x, cos.(x), label="cos(x)")

PyPlot.xlabel("x")
PyPlot.ylabel("y")
PyPlot.legend()
PyPlot.grid(true)
```
